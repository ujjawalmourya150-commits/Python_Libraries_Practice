```python
# ==========================================
# DAY 1 - PANDAS BASIC PRACTICE (10 QUESTIONS)
# ==========================================

import pandas as pd

# Q1. Create a Pandas Series
# Create a Series containing numbers from 10 to 50.
# Print the Series and its data type.

data = pd.Series([10, 20, 30, 40, 50])

print(data)
print(data.dtype)
```

```python
# Q2. Create a Series with Custom Index
# Create a Series containing marks: 85, 90, 75, 95, 80.
# Assign indexes: A, B, C, D, E.
# Print the Series.

data=pd.Series([85, 90, 75, 95, 80],index=['A','B','C','D','E'])
print(data)
```

```python
# Q3. Create a Series from a Dictionary
# Create a dictionary containing:
# Name: Rahul, Priya, Aman, Neha
# Marks: 85, 90, 78, 92
# Convert the dictionary into a Pandas Series.
# Print the result.

d={"Rahul":85, "Priya":90, "Aman":78, "Neha":92}
s=pd.Series(d)
print(s)
```

```python
# Q4. Create a DataFrame
# Create a DataFrame using the following data:
# Name: Rahul, Priya, Aman, Neha, Rohit
# Age: 20, 21, 19, 22, 20
# Marks: 85, 90, 78, 92, 88
d={"Name": ["Rahul", "Priya", "Aman", "Neha", "Rohit"],
    "Age":[ 20, 21, 19, 22, 20],
    "Marks":[85, 90, 78, 92, 88]}
df1=pd.DataFrame(d)
print(s)
```

```python
# Q5. Create a DataFrame from a Dictionary
# Create a dictionary with three columns:
# Name: A, B, C, D
# City: Indore, Pune, Mumbai, Delhi
# Salary: 25000, 30000, 28000, 35000
# Convert it into a DataFrame.

d={"Name": ['A', 'B', 'C', 'D' ],
   "City": ["Indore", "Pune", "Mumbai", "Delhi"],
   "Salary":[ 25000, 30000, 28000, 35000]}

df=pd.DataFrame(d)
print(df)
```

```python
# Q6. Display DataFrame Information
# Using the DataFrame created in Q4:
# 1. Display the first 3 rows.
# 2. Display the last 2 rows.
# 3. Display the shape.
# 4. Display the column names.
# 5. Display the data types of all columns.

print(df1.head(3))
print(df1.tail(2))
print(df1.shape)
print(df1.columns)
print(df1.dtypes)
```
