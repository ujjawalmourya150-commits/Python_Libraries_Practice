```python
import pandas as pd

data = pd.read_csv("Advertising1.csv")


# Q1. Display only the TV column from the dataset.


# Q2. Display only the TV and Sales columns.


# Q3. Display the first 10 rows of TV and Radio columns.


# Q4. Find the total amount spent on TV advertising.


# Q5. Calculate the average sales.


# Q6. Find the maximum and minimum values of the Sales column.


# Q7. Display all rows where Sales is greater than 15.


# Q8. Display all rows where TV advertising expenditure is greater than 100.


# Q9. Display rows where Sales is greater than 15 AND TV expenditure is greater than 200.


# Q10. Sort the dataset according to Sales in descending order
# and display the top 10 rows.
```

```python
# Q1. Display only the TV column from the dataset.

print(data.columns)
print(data['TV'])
```

```python
# Q2. Display only the TV and Sales columns.

print(data[["TV","Sales"]])
```

```python
# Q3. Display the first 10 rows of TV and Radio columns.

print(data[['TV','Radio']].head())
```

```python
# Q4. Find the total amount spent on TV advertising.

print(data['TV'].sum())
```

```python
# Q5. Calculate the average sales.

print(data['Sales'].mean())
```

```python
# Q6. Find the maximum and minimum values of the Sales column.

print(data['Sales'].agg(['min', 'max']))
```

```python
# Q7. Display all rows where Sales is greater than 15.

print(data[data['Sales']>15])
```

```python
# Q8. Display all rows where TV advertising expenditure is greater than 100.

print(data[data['TV']>100])
```

```python
# Q9. Display rows where Sales is greater than 15 AND TV expenditure is greater than 200.

print(data[(data['Sales'] > 15) & (data['TV'] > 200)])
```

```python
# Q10. Sort the dataset according to Sales in descending order
# and display the top 10 rows.

print(data.sort_values(by='Sales', ascending=False).head(10))
```
