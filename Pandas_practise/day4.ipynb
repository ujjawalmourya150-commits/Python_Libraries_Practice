```python
import pandas as pd

data=pd.read_csv("Advertising1.csv")
```

```python
# Q1. Row and Column Slicing
# Display rows from index 20 to 50 (inclusive)
# and only the TV, Radio and Sales columns.

print(data.loc[20:50, ['TV', 'Radio', 'Sales']])
```

```python
# Q2. Label-Based Selection (loc)
# Using loc, display rows from index 30 to 60 (inclusive)
# and only the Radio and Sales columns.

print(data.loc[30:60,['Radio','Sales']])
```

```python
# Q3. Position-Based Selection (iloc)
# Display rows at positions 10 to 30 (inclusive)
# and only the first 3 columns.

print(data.iloc[10:31,0:3])
```

```python
# Q4. Conditional Selection with loc
# Using loc, display all rows where:
# Sales is greater than 15 AND Radio is greater than 30.
# Display only TV, Radio and Sales columns.

print(data.loc[(data['Sales'] > 15) & (data['Radio'] > 30), ['TV', 'Radio', 'Sales']])
```

```python
# Q5. Multiple Column Sorting
# Sort the dataset:
# First by Sales in descending order.
# If Sales values are equal, sort by TV in ascending order.
# Display the top 15 rows.

print(data.sort_values(by=['Sales','TV'],ascending=[False,True]).head(15))
```

```python
# Q6. Slicing After Sorting
# Sort the complete dataset by TV in ascending order.
# Display rows from position 20 to 39 of the sorted dataset.

print(data.sort_values(by='TV').iloc[20:40])
```

```python
# Q7. loc with Multiple Conditions
# Using loc, select rows where:
# TV is greater than 250 OR Newspaper is greater than 100.
# Display only Radio, Newspaper and Sales columns.

print(data.loc[(data['TV']>250) | (data['Newspaper']>100),['Radio','Newspaper','Sales']])
```

```python
# Q8. iloc with Negative Indexing
# Display the last 20 rows of the dataset.
# Select only the last 3 columns.

print(data.iloc[-20:, -3:])
```

```python
# Q9. Sorting + Conditional Filtering
# Find all rows where:
# Sales is greater than 18 AND TV is greater than 200.
# Sort the resulting rows by Sales in descending order.
# Display only the top 10 rows.

print(data[(data['Sales']>18) & (data['TV']>200)].sort_values(by='Sales',ascending=False).head(10))
```

```python
# Q10. Advanced Challenge
# Perform the following operations:
# 1. Select rows where Radio is greater than 25.
# 2. Sort those rows by Sales in descending order.
# 3. From the sorted result, select rows at positions 5 to 14.
# 4. Display only TV, Radio and Sales columns.
# Use loc and iloc wherever appropriate.

print(data[data['Radio']>25].sort_values(by='Sales',ascending=False).iloc[5:15][['TV','Radio','Sales']])
```
